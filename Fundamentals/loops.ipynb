For loop

```python
for i in range(5):
    print(i**2)
```

```python
for i in range(1,10):
    if i%2 == 0:
        print(i)
```

```python
n = 5
for i in range(1,n+1):
    print('*'*i)
```

```python
for i in range(1,n+1):
    print('*'*(n+1-i))
```

```python
for i in range(1,n+1):
    print(' '*(n-i)+'*'*(2*i - 1))
```

```python
for i in range(n, 0, -1):
    spaces = " " * (n - i)
    stars = "*" * (2 * i - 1)
    print(spaces + stars)
```
