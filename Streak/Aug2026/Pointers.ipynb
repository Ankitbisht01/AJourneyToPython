```python

#Is Anagram

def is_anagram(s , t):
    #s = str(num)
    if len(s) != len(t):
        return False

    return sorted(s) == sorted(t)
    

is_anagram('bata','aatb')
```

```python

#Is Anagram

def is_anagram(s , t):
    #s = str(num)
    s_map = {}
    t_map = {}

    if len(s) != len(t):
        return False

    for char in s:
        if char in s_map:
            s_map[char] += 1

        else:
            s_map[char] = 1

    for char in t:
        if char in t_map:
            t_map[char] += 1
        else:
            t_map[char] = 1

    return s_map == t_map

     

is_anagram('bata','tvba')
```
