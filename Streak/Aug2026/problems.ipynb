**Problem 1: Word Frequency Counter**
Given a string text, return a dictionary containing the frequency of each unique word. The matching should be case-insensitive, and words are separated by single spaces.

Input: text = "The quick brown fox jumps over the lazy dog"
Output: {"the": 2, "quick": 1, "brown": 1, "fox": 1, "jumps": 1, "over": 1, "lazy": 1, "dog": 1}

```python
#Standard loop
def word_frequency_basic(text: str) -> dict[str, int]:
    # Convert string to lowercase and split into words
    words = text.lower().split()
    #print(words)
    freq = {}

    for word in words:
        if word not in freq:
            freq[word] = 1

        else:
            freq[word] = freq[word]+1

    return freq

    

# Example Usage
text = "The quick brown fox jumps over the lazy dog and fox"
print(word_frequency_basic(text))
# Output: {'the': 2, 'quick': 1, 'brown': 1, 'fox': 1, 'jumps': 1, 'over': 1, 'lazy': 1, 'dog': 1}
```

```python
#Using dict.get() Method
def word_frequency_get(text: str) -> dict[str, int]:
    words = text.lower().split()
    freq = {}

    for word in words:
        freq[word] = freq.get(word, 0) + 1

    return freq

text = "The quick brown fox jumps over the lazy dog and fox"
print(word_frequency_basic(text))
```

```python
#Using Counter
from collections import Counter


def word_frequency_counter(text: str) -> dict[str, int]:
    # Counter automatically counts occurrences of elements in an iterable
    return dict(Counter(text.lower().split()))

text = "The quick brown fox jumps over the lazy dog and fox"
print(word_frequency_basic(text))
```

**Problem 2: First Non-Repeating Character**
Given a string s, find the first non-repeating character in it and return its index. If it does not exist, return -1

Input: s = "leetcode"
Output: 0
Explanation: 'l' is the first character that appears only once.

```python
def firstUniqChar_hashmap(s: str) -> int:
    counts = {}

    # Pass 1: Build frequency map
    for char in s:
        counts[char] = counts.get(char, 0) + 1

    # Pass 2: Find the first character with a count of 1
    for index, char in enumerate(s):
        if counts[char] == 1:
            return index

    return -1


# Example Usage
print(firstUniqChar_hashmap("leetcode"))  # Output: 0
print(firstUniqChar_hashmap("loveleetcode"))  # Output: 2
print(firstUniqChar_hashmap("aabb"))  # Output: -1

```

**Problem 3: Group Students by Grade**
Given a list of tuples records where each tuple contains (student_name, grade), return a dictionary grouping student names under their respective grade key.

Input: records = [("Alice", "A"), ("Bob", "B"), ("Charlie", "A"), ("David", "C")]
Output: {"A": ["Alice", "Charlie"], "B": ["Bob"], "C": ["David"]}

```python
def group_students_basic(records: list[tuple[str, str]]) -> dict[str, list[str]]:
    grouped = {}

    for name, grade in records:
        if grade not in grouped:
            grouped[grade] = []
        grouped[grade].append(name)

    return grouped


# Example Usage
records = [("Alice", "A"), ("Bob", "B"), ("Charlie", "A"), ("David", "C")]
print(group_students_basic(records))
# Output: {'A': ['Alice', 'Charlie'], 'B': ['Bob'], 'C': ['David']}
```
