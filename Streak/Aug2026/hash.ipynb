**Two Sum Problem**

```python
#Brute force:
def two_sum(nums, target):
    n = len(nums)
    for i in range(n-1):  #index(0,1,2,3,4)

        for j in range(i+1, n):
             
             if nums[i]+nums[j] == target:
                 return i,j
    return None
           

print(two_sum([2,7,11,15],9))  # -> [0, 1]




```

```python
#Hash map

def two_sum(nums,target):
    seen = {}

    for i, num in enumerate(nums):
        comp = target - num
        if comp in seen:
            return seen[comp],i 
        else:
            seen[num] = i


        
    return None

print(two_sum([3,2,4],5))  # -> [0, 1]]







```
