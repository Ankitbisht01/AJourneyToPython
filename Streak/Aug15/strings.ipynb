```python
nums = [0,0,1,1,1,2,3,3,3,3,4]

left,right = 0,1
ans = []


while right < len(nums):
    if nums[left] == nums[right]:
        right += 1
    else:
        ans.append(nums[left])
        left = right
        right += 1

ans.append(nums[left])
nums = ans

print(ans)
print(len(ans))
print(nums)
```
